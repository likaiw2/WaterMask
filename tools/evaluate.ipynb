```python
import os
import numpy as np
from PIL import Image

categories_UIIS_to_SUIM=(0,6,5,2,3,1,4,7)
categories_SUIM_to_UIIS=(0,5,3,4,6,2,1,7)

HD_IOU=[]   # 001 1 in SUIM     
WR_IOU=[]   # 011 3 in SUIM
RO_IOU=[]   # 100 4 in SUIM
RI_IOU=[]   # 101 5 in SUIM
FV_IOU=[]   # 110 6 in SUIM

gt_image_root = "/home/liw324/code/data/SUIM_datasets/SUIM/TEST/images"
gt_mask_root = "/home/liw324/code/data/SUIM_datasets/SUIM/TEST/masks"
pred_mask_root = "/home/liw324/code/WaterMask/out_SUIM/20241016_173450_water_r50_fpn_1x/eval/water_r50_fpn_1x/pred_mask"

# gt_image_root = "/home/liw324/code/data/SUIM_datasets/SUIM/TEST/images"
# gt_mask_root = "/home/liw324/code/data/SUIM_datasets/SUIM/TEST/masks"
# pred_mask_root = "/home/liw324/code/WaterMask/out_SUIM/20241016_173450_water_r50_fpn_1x/eval/water_r50_fpn_1x/pred_mask"

# for item in os.listdir(gt_mask):
    
```

```python
# calculate HD IOU      001 1 in SUIM
for image in os.listdir(gt_image_root):
    image_name = image[:-4]
    gt_mask_path = f"{gt_mask_root}/HD/{image_name}.bmp"
    pred_mask_path = f"{pred_mask_root}/{image_name}_1.bmp"
    
    try:
        gt_mask = np.array(Image.open(gt_mask_path))/255
        # print(np.unique(gt_mask))
        pred_mask = np.array(Image.open(pred_mask_path))
        pred_mask = pred_mask.max(axis=2)/255
        # print(np.unique(pred_mask))
        
    except FileNotFoundError:
        continue
    
    sum=gt_mask+pred_mask
    # print(np.unique(sum))
    
    count_1 = np.count_nonzero(sum == 1)
    count_2 = np.count_nonzero(sum == 2)
    
    HD_IOU.append(count_2/(count_1+count_2))
    
    # print(gt_mask.shape,"    ",gt_mask_path)
    # print(pred_mask.shape,"    ",pred_mask_path)
print(HD_IOU)
    
```
